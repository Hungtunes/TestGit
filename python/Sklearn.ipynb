```python
import time
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score
import numpy as np

# 1. Đọc dữ liệu
start = time.time()
print("Đang tiến hành đọc file csv ...")

file_path = r"E:\Code\NaiveBayes\Dataset\datafinal.csv"
df = pd.read_csv(file_path)

print("Đã đọc file thành công!")
end = time.time()
print("Thời gian đọc file csv:", end - start, "s")

# 2. Tách features và label
start = time.time()

y = df[" Label"]           # giống df.label(" Label") trong C++
X = df.drop(columns=[" Label"])

end = time.time()
print("Thời gian tách features/label:", end - start, "s")

# 3. Train-test split
start = time.time()
print("Tiến hành tách train test ...")

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.30, shuffle=True, random_state=42
)

print("Size X_train:", len(X_train))
print("Size X_test:", len(X_test))

end = time.time()
print("Thời gian tách train, test:", end - start, "s")

# 4. Chuẩn hóa
start = time.time()
print("\nTiến hành chuẩn hóa ...")

scaler = StandardScaler()
scaler.fit(X_train)

X_train = scaler.transform(X_train)
X_test = scaler.transform(X_test)

end = time.time()
print("Thời gian chuẩn hóa:", end - start, "s")

# 5. Huấn luyện GaussianNB
start = time.time()
print("Training model...")

model = GaussianNB()
model.fit(X_train, y_train)

end = time.time()
print("Thời gian huấn luyện:", end - start, "s")

# 6. Dự đoán
start = time.time()
print("Predicting...")
y_pred = model.predict(X_test)
end = time.time()
print("Thời gian dự đoán:", end - start, "s")


# 7. Đánh giá
# === Classification Report ===
print("\n=== Classification Report ===")
print(classification_report(y_test, y_pred))

# === Confusion Matrix Đẹp (Dùng Pandas) ===
cm = confusion_matrix(y_test, y_pred)
labels = sorted(y.unique())

# Tạo DataFrame từ confusion matrix
# Index (hàng) là True label, Columns (cột) là Predicted label
df_cm = pd.DataFrame(
    cm, 
    index=[f"True-{label}" for label in labels], 
    columns=[f"Pred-{label}" for label in labels]
)

# Cấu hình hiển thị Pandas để không bị xuống dòng nếu bảng quá rộng
pd.set_option('display.max_columns', None)  # Hiển thị hết các cột
pd.set_option('display.width', 1000)        # Mở rộng chiều ngang hiển thị
pd.set_option('display.colheader_justify', 'center') # Căn giữa tiêu đề

print("\n=== Confusion Matrix ===")
print(df_cm)

# === Accuracy chung ===
acc = accuracy_score(y_test, y_pred)
print("\nAccuracy:", round(acc * 100, 2), "%")
```

```python
import time
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score
import numpy as np

# 1. Đọc dữ liệu
start = time.time()
print("Đang tiến hành đọc file csv ...")

file_path = r"E:\Code\NaiveBayes\Dataset\datafinal.csv"
df = pd.read_csv(file_path)

print("Đã đọc file thành công!")
end = time.time()
print("Thời gian đọc file csv:", end - start, "s")

# 2. Tách features và label
start = time.time()

y = df[" Label"]
X = df.drop(columns=[" Label"])

end = time.time()
print("Thời gian tách features/label:", end - start, "s")

# 3. Train-test split
start = time.time()
print("Tiến hành tách train test ...")

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.30, shuffle=True, random_state=42
)

print("Size X_train:", len(X_train))
print("Size X_test:", len(X_test))

end = time.time()
print("Thời gian tách train, test:", end - start, "s")

# 4. Chuẩn hóa (Random Forest không bắt buộc nhưng vẫn giữ giống C++)
start = time.time()
print("\nTiến hành chuẩn hóa ...")

scaler = StandardScaler()
scaler.fit(X_train)

X_train = scaler.transform(X_train)
X_test = scaler.transform(X_test)

end = time.time()
print("Thời gian chuẩn hóa:", end - start, "s")

# 5. Huấn luyện Random Forest
start = time.time()
print("Training model...")

model = RandomForestClassifier(
    n_estimators=200,      # số cây
    max_depth=None,        # độ sâu không giới hạn
    random_state=42,
    n_jobs=-1              # dùng tất cả CPU để chạy nhanh
)
model.fit(X_train, y_train)

end = time.time()
print("Thời gian huấn luyện:", end - start, "s")

# 6. Dự đoán
start = time.time()
print("Predicting...")
y_pred = model.predict(X_test)
end = time.time()
print("Thời gian dự đoán:", end - start, "s")

# 7. Đánh giá
# === Classification Report ===
print("\n=== Classification Report ===")
print(classification_report(y_test, y_pred))

# === Confusion Matrix Đẹp (Dùng Pandas) ===
cm = confusion_matrix(y_test, y_pred)
labels = sorted(y.unique())

# Tạo DataFrame từ confusion matrix
# Index (hàng) là True label, Columns (cột) là Predicted label
df_cm = pd.DataFrame(
    cm, 
    index=[f"True-{label}" for label in labels], 
    columns=[f"Pred-{label}" for label in labels]
)

# Cấu hình hiển thị Pandas để không bị xuống dòng nếu bảng quá rộng
pd.set_option('display.max_columns', None)  # Hiển thị hết các cột
pd.set_option('display.width', 1000)        # Mở rộng chiều ngang hiển thị
pd.set_option('display.colheader_justify', 'center') # Căn giữa tiêu đề

print("\n=== Confusion Matrix ===")
print(df_cm)

# === Accuracy chung ===
acc = accuracy_score(y_test, y_pred)
print("\nAccuracy:", round(acc * 100, 2), "%")
```
